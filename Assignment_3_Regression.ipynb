```python
# Importing required libraries
import pandas as pd
import numpy as np
from sklearn.datasets import fetch_california_housing
from sklearn.preprocessing import StandardScaler

# Load the dataset
housing = fetch_california_housing()

# Convert to pandas DataFrame
df = pd.DataFrame(housing.data, columns=housing.feature_names)
df['MedHouseVal'] = housing.target  # Add the target column to the DataFrame

# Display the first few rows
df.head()
```

```python
# Check for missing values
df.isnull().sum()
```

Feature Scaling (Standardization)

```python
from sklearn.preprocessing import StandardScaler

# Original features and target
X = df.drop('MedHouseVal', axis=1)
y = df['MedHouseVal']

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## Data Loading and Preprocessing

We used `fetch_california_housing` from `sklearn.datasets` to load the California Housing dataset, which includes various features such as median income, average number of rooms, population, etc.

We converted the dataset into a pandas DataFrame for easier data handling and added the target variable `MedHouseVal` which represents the median house value.

### Missing Values
We checked for missing values using `isnull().sum()` and found no missing data, so no imputation was necessary.

### Feature Scaling
We applied StandardScaler from sklearn to standardize the features. This ensures all features have a mean of 0 and standard deviation of 1, which is especially important for algorithms like SVR and Gradient Boosting that are sensitive to feature magnitudes.


## Model Training

```python
# Import regression models
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR

# Initialize models
models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(random_state=42),
    "Gradient Boosting": GradientBoostingRegressor(random_state=42),
    "SVR": SVR()
}

# Train models
trained_models = {}
for name, model in models.items():
    model.fit(X_scaled, y)
    trained_models[name] = model
```

## Regression Algorithms

We implemented five regression models to predict the median house value (`MedHouseVal`) using various housing features:

1. **Linear Regression**:
   - Assumes a linear relationship between input features and the target.
   - Suitable for quick baseline modeling, but may underperform with non-linear data.

2. **Decision Tree Regressor**:
   - Splits the data into branches to minimize prediction error.
   - Can capture non-linear relationships but may overfit on small datasets.

3. **Random Forest Regressor**:
   - An ensemble of decision trees that improves prediction accuracy and reduces overfitting.
   - Works well with high-dimensional data and provides better generalization.

4. **Gradient Boosting Regressor**:
   - Builds trees sequentially, each trying to correct the errors of the previous one.
   - Often provides high accuracy but is slower to train.

5. **Support Vector Regressor (SVR)**:
   - Uses a margin of tolerance (epsilon) and finds a hyperplane that fits within this margin.
   - Sensitive to feature scaling and works well on small-to-medium datasets.


## Model Evaluation and Comparison

Code to Evaluate All Models

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Store evaluation results
results = []

# Evaluate each model
for name, model in trained_models.items():
    y_pred = model.predict(X_scaled)
    
    mse = mean_squared_error(y, y_pred)
    mae = mean_absolute_error(y, y_pred)
    r2 = r2_score(y, y_pred)
    
    results.append({
        'Model': name,
        'MSE': mse,
        'MAE': mae,
        'R² Score': r2
    })

# Convert to DataFrame for easy viewing
results_df = pd.DataFrame(results)
results_df.sort_values(by='R² Score', ascending=False)
```

## Model Evaluation and Comparison

We evaluated each regression model using:
- **Mean Squared Error (MSE)**: Penalizes larger errors more heavily.
- **Mean Absolute Error (MAE)**: Averages absolute differences between actual and predicted values.
- **R² Score**: Represents the proportion of variance explained by the model.

### Comparison:
- **Best Model**: Random Forest Regressor — It had the highest R² score and lowest error metrics, showing strong performance on this dataset.
- **Worst Model**: SVR — While useful on smaller datasets, it underperformed here, likely due to the larger dataset size and its sensitivity to parameter tuning and scale.


# 📊 Summary & Conclusion

In this assignment, we applied five regression algorithms to the California Housing dataset:

- **Best Model**: The **Random Forest Regressor** achieved the highest R² score, indicating strong predictive performance.
- **Worst Model**: The **Support Vector Regressor** underperformed, likely due to its sensitivity to parameter tuning and scale in larger datasets.
